```python
import numpy as np

# Generate example data inspired by
# https://vita.had.co.nz/papers/boxplots.pdf (Figure 4)

rng = np.random.default_rng(42)
N = 1000

n  = rng.standard_normal(N)                            # standard normal
s  = rng.lognormal(0, 0.5, N) - 1.0                    # right-skewed (shifted lognormal)
k  = rng.standard_t(df=4, size=N) * 0.5                # leptokurtic (heavy-tailed t)
mm = np.concatenate([rng.normal(-1, 0.35, N // 2),     # bimodal mixture
                     rng.normal( 1, 0.35, N // 2)])
rng.shuffle(mm)

datasets = [mm, k, s, n]
labels = ['bimodal', 'leptokurtic', 'right-skewed', 'normal']
```

```python
import pavement.matplotlib as pavement

_ = pavement.plot(datasets, labels=labels, bins=16, orientation='horizontal')
```

```python
import pavement.plotly as pavement

pavement.plot(datasets, labels=labels, bins=16, orientation='horizontal').show()
```

```python
import pavement.bokeh as pavement
from bokeh.io import output_notebook
from bokeh.plotting import show

output_notebook(hide_banner=True)
show(pavement.plot(datasets, labels=labels, bins=16, orientation='horizontal'))
```

```python
import pavement.holoviews as pavement
import holoviews as hv

hv.extension('bokeh')
pavement.plot(datasets, labels=labels, bins=16, orientation='horizontal')
```

### Rug plots

As in the seaborn [docs](https://seaborn.pydata.org/generated/seaborn.rugplot.html)

![seaborn rug](https://seaborn.pydata.org/_images/rugplot_3_0.png)

```python
import seaborn as sns
# sns.set_theme()  # skip to keep other plots consistent
tips = sns.load_dataset("tips")

import pavement.matplotlib as pavement
import matplotlib.pyplot as plt
```

```python
ax = sns.scatterplot(data=tips, x="total_bill", y="tip")
props = {"color": "#4c72b0"}
pavement.margin(tips.total_bill, axis="x", bins=16, where='inside bottom', line_props=props)
_ = pavement.margin(tips.tip, axis="y", bins=16, where='inside left', line_props=props)
```

```python
ax = sns.scatterplot(data=tips, x="total_bill", y="tip")
props = {"color": "#4c72b0", "alpha": 0.3}
pavement.margin(tips.total_bill, axis="x", bins=None, where='inside bottom', line_props=props)
_ = pavement.margin(tips.tip, axis="y", bins=None, where='inside left', line_props=props)
```

```python
import pandas as pd
import matplotlib.ticker as mticker

incomes = pd.read_csv('incomes.csv')

pavement.plot2d(x=incomes.dollars, y=incomes.age, weights=incomes.sample_weight, bins=10, first_split='y')
plt.xlim(-10_000, 185_000)
plt.gca().xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${int(x/1000)}k' if x != 0 else '$0'))
plt.xlabel('2024 personal income (min and deciles shown; max off right)')
plt.ylabel('Age (deciles for income earners 20-69 years old)')
plt.title('Distribution of personal income based on 2024 ACS')
plt.show()
```
